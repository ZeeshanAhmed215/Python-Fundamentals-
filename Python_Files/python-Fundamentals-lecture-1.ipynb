#Q1. Write a program to input 2 numbers and print their sum

```python
num1=input("Enter first number: ")
num2=input("Enter second number: ")
print("The sum  of ", num1, "and", num2, "is", int(num1)+int(num2))
```

Q2. Write a program to input the side of a square and print its
area.

```python
num1 = input("Enter side of square: ")

print(f"Area of side {num1} is: {int(num1) * int(num1)}")
```

Q3. Write a program to input 2 floating-point numbers and
print their average.

```python
num1=float(input("Enter number:"))
num2=float(input("Enter number:"))
sum_num=num1+num2
ave=sum_num/2
print(f"Average of {num1} and {num2} is {ave}")
```

Q4. Input 2 integers a and b. Print True if a is greater than or
equal to b, otherwise print False.

```python
a=int(input("Enter a:"))
b=int(input("Enter b:"))
if a>=b:
    print(True)
else:
    print(False)
```
